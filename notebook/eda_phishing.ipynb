<h4>Class Distribution of Target Variable </h4>

```python
target_col = "phishing"

# Count of each class
df[target_col].value_counts()
```

```python
# Percentage of each class
df[target_col].value_counts(normalize=True) * 100
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(x=target_col, data=df)
plt.title("Class Distribution of Phishing Website Dataset")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

<h4>Correlation Heatmap</h4>

```python
plt.figure(figsize=(18, 12))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, cmap="coolwarm", linewidths=0.2)
plt.title("Correlation Heatmap")
plt.show()
```

<h4>Correlation with Target Column</h4>

```python
corr_with_target = df.corr(numeric_only=True)[target_col].sort_values(ascending=False)
corr_with_target
```

<h4>URL Length Analysis</h4>

```python
plt.figure(figsize=(8,5))
sns.histplot(df["length_url"], bins=30, kde=True)
plt.title("Distribution of URL Length")
plt.xlabel("URL Length")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(8,4))
sns.boxplot(y=df["length_url"])
plt.title("Boxplot of URL Length")
plt.ylabel("URL Length")
plt.show()
```

<h4>Special Character Analysis</h4>

```python
special_cols = [
    "qty_at_url",
    "qty_hyphen_url",
    "qty_questionmark_url",
    "qty_equal_url",
    "qty_percent_url",
    "qty_slash_url"
]

plt.figure(figsize=(10,5))
df[special_cols].mean().plot(kind="bar")
plt.title("Average Special Character Usage in URL")
plt.xlabel("Features")
plt.ylabel("Average Value")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

<h4>Shortened URL Analysis</h4>

```python
plt.figure(figsize=(6,4))
sns.countplot(x="url_shortened", data=df)
plt.title("Distribution of Shortened URL Feature")
plt.xlabel("Shortened URL")
plt.ylabel("Count")
plt.show()
```

<h4>Redirect Count Analysis</h4>

```python
plt.figure(figsize=(8,5))
sns.histplot(df["qty_redirects"], bins=30, kde=True)
plt.title("Distribution of Redirect Count")
plt.xlabel("Redirect Count")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(8,4))
sns.boxplot(y=df["qty_redirects"])
plt.title("Boxplot of Redirect Count")
plt.ylabel("Redirect Count")
plt.show()
```

<h4>Feature Comparison with Target</h4>

```python
features_to_compare = ["length_url", "qty_at_url", "qty_redirects"]

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

for ax, col in zip(axes, features_to_compare):
    sns.boxplot(x="phishing", y=col, data=df, ax=ax)
    ax.set_title(f"{col} vs Target")
    ax.set_xlabel("Target")
    ax.set_ylabel(col)

plt.tight_layout()
plt.show()
```
